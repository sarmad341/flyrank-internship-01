## 5. Why ML beats a fixed rule here

From Notebook 02, the hand rule (stale × visible) and a shallow decision tree traded wins depending on the metric and split the hand rule won Precision@20 (0.900 vs 0.550) but results flipped at Precision@50 depending on the run. That instability is itself the evidence: the boundary between "worth reviewing" and "not" isn't a single clean threshold on one or two variables, it's a joint pattern across several signals (impressions_90d, content_age_days, avg_position, days_since_last_update) that interact in ways a single if-statement can't capture. An if-statement also can't be systematically re-tuned as those interactions shift over time the way a retrained model can.
