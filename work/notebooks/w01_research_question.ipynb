## 2. The question: decision, action, cost of a wrong call

What decision does your work improve?

Which content pages a review team should prioritize refreshing first, given limited time and review capacity.

Who acts on it?

A content strategist or reviewer working through a ranked queue — they take the top of the list and start reviewing/refreshing those pages first.

What does a wrong recommendation cost?

Two-sided cost. A false positive (flagging a page that isn't actually a priority) wastes the reviewer's limited time on a page that didn't need attention. A false negative (missing a page that's genuinely declining) means that page keeps losing visibility, unaddressed, until the next review cycle. Because review capacity is fixed and small relative to the full content inventory, precision at the top of the ranked list (Precision@K) matters more than overall accuracy getting the top 20-50 right matters more than being right on average across all pages.
